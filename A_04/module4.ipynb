# Assignment 04: Multilayer Perceptron (MLP) Implementation

## Overview
This notebook demonstrates the implementation of a **Multilayer Perceptron** with forward propagation across 4 sequential layers:
* **Input Layer:** 3 neurons
* **Hidden Layer 1:** 5 neurons (Sigmoid activation)
* **Hidden Layer 2:** 5 neurons (Sigmoid activation)
* **Hidden Layer 3:** 2 neurons (Sigmoid activation)
* **Output Layer (Layer 4):** 2 neurons (ReLU activation)

```python
import numpy as np
from multilayer_perceptron import MultilayerPerceptron

# Instantiate the MLP model and define sample input
mlp = MultilayerPerceptron()
x = np.array([0.5, -1.2, 0.8])

print("Input Vector:", x)
results = mlp.forward(x)
```

## Forward Propagation Results
Below are the computed activations across each hidden layer and the final output layer:

```python
print("Hidden Layer 1 Output:", np.round(results["hidden_1"], 4))
print("Hidden Layer 2 Output:", np.round(results["hidden_2"], 4))
print("Hidden Layer 3 Output:", np.round(results["hidden_3"], 4))
print("Output Layer 4 Output:", np.round(results["output_layer_4"], 4))
```
