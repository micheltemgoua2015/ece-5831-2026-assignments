# Assignment 03: Logic Gate Implementation using Perceptrons and NumPy

This notebook demonstrates the implementation and evaluation of basic logic gates (AND, NAND, OR, NOR, XOR) using single-layer and multi-layer perceptron models with NumPy.

```python
import numpy as np
from logic_gate import LogicGate
```

## 1. Class Instantiation & Testing Setup
We initialize an instance of the `LogicGate` class to evaluate all logic gates across truth tables.

```python
gate = LogicGate()
test_inputs = [(0, 0), (0, 1), (1, 0), (1, 1)]
```

## 2. AND Gate Evaluation

```python
for x1, x2 in test_inputs:
    gate.and_gate(x1, x2)
    gate.print_output("AND")
```

## 3. NAND Gate Evaluation

```python
for x1, x2 in test_inputs:
    gate.nand_gate(x1, x2)
    gate.print_output("NAND")
```

## 4. OR Gate Evaluation

```python
for x1, x2 in test_inputs:
    gate.or_gate(x1, x2)
    gate.print_output("OR")
```

## 5. NOR Gate Evaluation

```python
for x1, x2 in test_inputs:
    gate.nor_gate(x1, x2)
    gate.print_output("NOR")
```

## 6. XOR Gate Evaluation
Multi-layer logic constructed by combining NAND, OR, and AND gates.

```python
for x1, x2 in test_inputs:
    gate.xor_gate(x1, x2)
    gate.print_output("XOR")
```
