# Assignment 03: Logic Gate Implementation using Perceptrons and NumPy

## Overview
This notebook demonstrates the implementation and evaluation of basic Boolean logic gates (**AND**, **NAND**, **OR**, **NOR**, and **XOR**) constructed using single-layer and multi-layer perceptron models with **NumPy**.

### Mathematical Model
Each single-layer gate computes a weighted sum using the vector dot product, followed by a step threshold activation:

```python
import numpy as np
import logic_gate as lg

# Initialize LogicGate instance and define truth table inputs
gate = lg.LogicGate()
test_inputs = [(0, 0), (0, 1), (1, 0), (1, 1)]
```

## 1. AND Gate Evaluation
The **AND** gate returns `1` only when both inputs \(x_1\) and \(x_2\) are `1`. 
* **Weights:** \(w_1 = 0.5, w_2 = 0.5\)
* **Threshold:** \(\theta = 0.99\)

```python
print("================ AND GATE TEST ================")
for x1, x2 in test_inputs:
    gate.and_gate(x1, x2)
    gate.print_output("AND")
```

## 2. NAND Gate Evaluation
The **NAND** (NOT-AND) gate is the inverse of the AND gate. It returns `0` only when both inputs are `1`.
* **Weights:** \(w_1 = -0.5, w_2 = -0.5\)
* **Threshold:** \(\theta = -0.70\)

```python
print("================ NAND GATE TEST ================")
for x1, x2 in test_inputs:
    gate.nand_gate(x1, x2)
    gate.print_output("NAND")
```

## 3. OR Gate Evaluation
The **OR** gate returns `1` if at least one input is `1`.
* **Weights:** \(w_1 = 0.5, w_2 = 0.5\)
* **Threshold:** \(\theta = 0.00\)

```python
print("================ OR GATE TEST ================")
for x1, x2 in test_inputs:
    gate.or_gate(x1, x2)
    gate.print_output("OR")
```

## 4. NOR Gate Evaluation
The **NOR** (NOT-OR) gate returns `1` only when both inputs are `0`.
* **Weights:** \(w_1 = -0.5, w_2 = -0.5\)
* **Threshold:** \(\theta = -0.10\)

```python
print("================ NOR GATE TEST ================")
for x1, x2 in test_inputs:
    gate.nor_gate(x1, x2)
    gate.print_output("NOR")
```

## 5. XOR Gate Evaluation
The **XOR** (Exclusive OR) gate is **non-linearly separable** and cannot be solved with a single linear threshold unit. It is constructed as a multi-layer network combining **NAND**, **OR**, and **AND** gates:

```python
print("================ XOR GATE TEST ================")
for x1, x2 in test_inputs:
    gate.xor_gate(x1, x2)
    gate.print_output("XOR")
```
